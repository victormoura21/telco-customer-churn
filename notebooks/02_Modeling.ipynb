# 02 — Modelagem

O holdout é separado antes da seleção. O `GridSearchCV` compara três classificadores somente no conjunto de treino e o holdout é avaliado após a seleção.

Os caminhos de saída são passados explicitamente a partir de `ROOT`. Assim, os artefatos são gravados em `reports/generated/` e `models/` na raiz do projeto, independentemente de o notebook estar sendo executado a partir da pasta `notebooks/`.

```python
from pathlib import Path
import sys
import pandas as pd

ROOT = Path.cwd()

# Quando o kernel inicia dentro de notebooks/, voltamos para a raiz do projeto.
if not (ROOT / "data").exists():
    ROOT = ROOT.parent

sys.path.insert(0, str(ROOT))

from src.modeling import run_pipeline

print("Raiz do projeto:", ROOT.resolve())

result = run_pipeline(
    ROOT / "data" / "raw" / "telco_customer_churn.xlsx",
    output_dir=ROOT / "reports" / "generated",
    model_dir=ROOT / "models",
)
```

## Seleção no treino

```python
display(result["selection"].round(4))
print("Selecionado:", result["selected_model"])
```

## Avaliação final no holdout

```python
pd.DataFrame([result["test_metrics"]]).round(4)
```

A comparação entre modelos deve considerar também a variabilidade entre folds. Diferenças pequenas nas médias não demonstram superioridade inequívoca.
